```python
from matplotlib.colors import to_rgb
import numpy as np
import matplotlib.pyplot as plt



# Arms (now includes original U)
arms = ["real_only", "synthR_plus_real", "synthU_plus_real", "synthUnew_plus_real", "synth_Uoverlays_plus_real"]

# Consistent colours across all plots
arm_colors = {
    "real_only": "C0",
    "synthR_plus_real": "C1",
    "synthU_plus_real": "C2",
    "synthUnew_plus_real": "C3",
    "synth_Uoverlays_plus_real": "C4"

}

metrics = ["test_ap_lift", "test_auroc"]

def sem(x):
    x = np.asarray(x, dtype=float)
    if x.size == 0:
        return np.nan
    if x.size == 1:
        return 0.0
    return x.std(ddof=1) / np.sqrt(x.size)

# bar charts with SEM
for metric in metrics:
    n_grid = sorted(df["n_real"].unique())
    x = np.arange(len(n_grid))
    bar_width = 0.2
    offsets = np.linspace(-(bar_width*1.5), (bar_width*1.5), num=len(arms))

    fig, ax = plt.subplots(figsize=(11,5))
    rng = np.random.default_rng(12345)  # reproducible dot jitter

    for i, arm in enumerate(arms):
        means, sems, per_group_vals = [], [], []
        for n_val in n_grid:
            vals = df[(df["arm"]==arm) & (df["n_real"]==n_val)][metric].to_numpy(dtype=float)
            vals = vals[np.isfinite(vals)]
            per_group_vals.append(vals)
            means.append(np.nanmean(vals) if vals.size else np.nan)
            sems.append(sem(vals))

        xpos = x + offsets[i]

        # bars with SEM whiskers
        ax.bar(
            xpos, means, width=bar_width, label=arm_map.get(arm, arm), alpha=0.9,
            yerr=sems, capsize=4, color=arm_colors[arm], edgecolor="black", linewidth=0.6
        )

        # overlay per-seed dots (same hue as bar but darker)
        base_rgb = np.array(to_rgb(arm_colors[arm]))
        darker_rgb = base_rgb * 0.7  # <1 = darker
        for j, vals in enumerate(per_group_vals):
            if vals.size == 0:
                continue
            jitter = (rng.random(vals.size) - 0.5) * (bar_width * 0.5)
            ax.scatter(np.full(vals.shape, xpos[j]) + jitter, vals,
                       s=18, alpha=0.8, linewidths=0, color=darker_rgb)

    ax.set_xticks(x)
    ax.set_xticklabels(n_grid)
    ax.set_xlabel("n_real")
    ax.set_ylabel(f"{metric}")
    ax.set_title(f"{metric} mean (bars) ± SEM (whiskers) + per-seed dots")
    ax.legend(ncol=2)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt


# comparisons to run (only vs real_only)
comparisons = [
    ("synthR_plus_real", "real_only", "Realistic + Real > Real"),
    ("synthU_plus_real", "real_only", "Cartoon + Real > Real"),
    ("synthUnew_plus_real", "real_only", "Low Detail + Real > Real"),
    ("synth_Uoverlays_plus_real", "real_only", "Overlays + Real > Real"),
]

all_stats = []

for n_real, group in df.groupby("n_real"):
    for metric in metrics:
        for armA, armB, label in comparisons:
            gA = group[group["arm"]==armA].sort_values("seed")
            gB = group[group["arm"]==armB].sort_values("seed")

            merged = pd.merge(gA, gB, on="seed", suffixes=("_A", "_B"))
            diffs = merged[f"{metric}_A"] - merged[f"{metric}_B"]

            # stats
            t_stat, p_ttest = stats.ttest_rel(merged[f"{metric}_A"], merged[f"{metric}_B"])
            w_stat, p_wilcoxon = stats.wilcoxon(diffs)

            all_stats.append({
                "n_real": n_real,
                "metric": metric,
                "comparison": label,
                "mean_diff": diffs.mean(),
                "t_p": p_ttest,
                "wilcoxon_p": p_wilcoxon
            })

stats_df = pd.DataFrame(all_stats)


```

```python
# ---- plotting ----
for metric in metrics:
    fig, axes = plt.subplots(1,2, figsize=(12,4), sharex=True)
    sub = stats_df[stats_df["metric"]==metric]

    # Graph 1: mean diff
    for comp in sub["comparison"].unique():
        cur = sub[sub["comparison"]==comp]
        axes[0].plot(cur["n_real"], cur["mean_diff"], marker="o", label=comp)
    axes[0].axhline(0, color="k", linestyle="--")
    axes[0].set_title(f"{metric} mean differences")
    axes[0].set_xlabel("n_real")
    axes[0].set_ylabel("mean diff")
    axes[0].legend()

    # Graph 2: raw p-values (log scale)
    for comp in sub["comparison"].unique():
        cur = sub[sub["comparison"]==comp]
        axes[1].plot(cur["n_real"], cur["t_p"], marker="o", label=comp)
    axes[1].axhline(0.05, color="r", linestyle="--", label="p=0.05")
    axes[1].set_yscale("log")
    axes[1].invert_yaxis()
    axes[1].set_title(f"{metric} paired t-test p-values")
    axes[1].set_xlabel("n_real")
    axes[1].set_ylabel("p-value (log scale)")
    axes[1].legend()

    plt.show()
```
